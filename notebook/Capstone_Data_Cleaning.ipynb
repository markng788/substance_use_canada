```python
#importing necessary libraries
#reading a CSV file using pandas

import pandas as pd

df = pd.read_csv('/content/cstads_2122_pumf.csv')
display(df.head())
```

Data Cleaning and Impute Missing Value

```python
#Calculate the number of missing values in each column

missing_values = df.isnull().sum()
print("Missing Values in Each Column:")
missing_values
```

```python
#Impute Mean for Missing Value

# Calculate and Impute the mean of SS_020 where the value is not equal to 2
mean_ss_020 = df.loc[df['SS_020'] != 2, 'SS_020'].mean()
df.loc[df['SS_020'] == 2, 'SS_020'] = mean_ss_020

# Calculate and Impute the mean of ALC_030 where the value is not equal to 3
mean_alc_030 = df.loc[df['ALC_030'] != 3, 'ALC_030'].mean()
df.loc[df['ALC_030'] == 3, 'ALC_030'] = mean_alc_030

# Calculate and Impute the mean of ALC_060 where the value is not equal to 2
mean_alc_060 = df.loc[df['ALC_060'] != 2, 'ALC_060'].mean()
df.loc[df['ALC_060'] == 2, 'ALC_060'] = mean_alc_060

#Impute means for these columns
columns_to_impute = ['PH_010', 'PH_020', 'PH_110', 'PH_120', 'PH_070', 'PH_080', 'CA_020', 'ALC_080', 'CAN_050']
for column in columns_to_impute:
  mean_value = df.loc[df[column] != 5, column].mean()
  df[column] = df[column].fillna(mean_value).replace(5, mean_value)

df.head()
```

```python
# prompt: Impute mode for missing values of the following columns: DVGENDER, DVURBAN, DVRES, SS_010, SS_030, SS_040, ALC_010, CAN_010, BUL_010, BUL_020, BUL_030, BUL_040, DVLAST30
# columns_to_impute median = ['DVTY1ST', 'DVTY2ST', 'DVAMTSMK', 'BEH_010', 'BEH_020', 'CAN_130', 'CAN_040', 'CAN_030', 'CAN_020', 'ALC_071', 'ALC_060', 'ALC_050', 'ALC_040', 'ALC_030', 'ALC_020', 'SC_010', 'TS_011', 'SS_020', 'GH_010', 'GH_020']

columns_mode = ['DVGENDER', 'DVURBAN', 'DVRES', 'SS_010', 'SS_030', 'SS_040', 'ALC_010', 'CAN_010', 'BUL_010', 'BUL_020', 'BUL_030', 'BUL_040', 'DVLAST30']
columns_median = ['DVTY1ST', 'DVTY2ST', 'DVAMTSMK', 'BEH_010', 'BEH_020', 'PH_010', 'PH_020', 'PH_070', 'PH_080', 'PH_110', 'PH_120', 'CAN_130', 'CA_020', 'CAN_050', 'ALC_080', 'CAN_040', 'CAN_030', 'CAN_020', 'ALC_071', 'ALC_060', 'ALC_050', 'ALC_040', 'ALC_030', 'ALC_020', 'SC_010', 'TS_011', 'SS_020', 'GH_010', 'GH_020']

for column in columns_mode:
    mode_value = df[column].mode()[0]  # Get the first mode in case of multiple modes
    df[column] = df[column].fillna(mode_value)

for column in columns_median:
    median_value = df[column].median()
    df[column] = df[column].fillna(median_value)
```

```python
# Calculate the number of missing values in each column
missing_values = df.isnull().sum()

# Display the results
print("Missing Values in Each Column:")
missing_values
```
