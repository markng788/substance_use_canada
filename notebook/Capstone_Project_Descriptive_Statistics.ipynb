#Descriptive Statistics

```python
# Provide profile of respondents - frequency count and percentage of PROVID, DVGENDER, DVURBAN, DVRES, GRADE, each variable with a table of columns of frequency count and percentage
# Create bar charts for each variable

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Function to generate profile table and bar chart for a given variable
def profile_variable(df, variable_name):
    # Frequency count and percentage
    counts = df[variable_name].value_counts()
    percentages = df[variable_name].value_counts(normalize=True) * 100

    # Create a DataFrame for the profile table
    profile_table = pd.DataFrame({'Frequency': counts, 'Percentage': percentages.round(2)})

    print(f"\nProfile of {variable_name}:")
    display(profile_table)

    # Create a bar chart
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=variable_name, order=counts.index)
    plt.title(f'Distribution of {variable_name}')
    plt.xlabel(variable_name)
    plt.ylabel('Frequency')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

# List of variables to profile
variables_to_profile = ['PROVID', 'DVGENDER', 'DVURBAN', 'DVRES', 'GRADE']

# Generate profile and bar chart for each variable
for var in variables_to_profile:
    if var in df.columns:
        profile_variable(df, var)
    else:
        print(f"Warning: Variable '{var}' not found in the DataFrame.")
```

```python
# Frequency count and percentage of background variables: GH_010, GH_020, PH_010, PH_020, PH_110, PH_120, PH_070, PH_080, CA_020, ALC_080, CAN_050, BEH_010, BEH_020, BUL_010, BUL_020, BUL_030, BUL_040
# for each variable, shows frequency count and percentage as a table

import pandas as pd
background_vars = ['GH_010', 'GH_020', 'PH_010', 'PH_020', 'PH_110', 'PH_120', 'PH_070', 'PH_080', 'CA_020', 'ALC_080', 'CAN_050', 'BEH_010', 'BEH_020', 'BUL_010', 'BUL_020', 'BUL_030', 'BUL_040']

for col in background_vars:
    print(f"\nFrequency and Percentage for {col}:")
    freq_table = pd.DataFrame({
        'Frequency': df[col].value_counts(),
        'Percentage': df[col].value_counts(normalize=True) * 100
    })
    print(freq_table)
    print("-" * 30)

```

#Substance Abuse Behavior - Smoking

```python
# frequency and percentage of the smoking variables: SS_010, SS_020, TS_011, SS_030, SS_040, SC_010, DVTY1ST, DVTY2ST, DVLAST30, DVAMTSMK
# Each variable should have its table with frequency and percentage
# Plot the percentage of each variable as horizontal bar chart

import pandas as pd
import matplotlib.pyplot as plt
smoking_vars = ['SS_010', 'SS_020', 'TS_011', 'SS_030', 'SS_040', 'SC_010', 'DVTY1ST', 'DVTY2ST', 'DVLAST30', 'DVAMTSMK']

for col in smoking_vars:
    print(f"\nFrequency and Percentage for {col}:")
    counts = df[col].value_counts()
    percentages = df[col].value_counts(normalize=True) * 100

    freq_table = pd.DataFrame({
        'Frequency': counts,
        'Percentage': percentages.round(2)
    })
    print(freq_table)
    print("-" * 30)

    # Plot percentage as horizontal bar chart
    plt.figure(figsize=(10, 6))
    percentages.sort_values().plot(kind='barh', color='skyblue')
    plt.title(f'Percentage Distribution of {col}')
    plt.xlabel('Percentage (%)')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```
